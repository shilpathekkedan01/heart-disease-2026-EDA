### HEART DISEASE RISK

#### This project analyzes patient health, lifestyle, and clinical data to identify patterns and factors associated with heart disease. Using Python, Pandas, Matplotlib, and Seaborn, the dataset is explored through statistical analysis and visualizations to understand risk factors such as age, blood pressure, cholesterol, BMI, smoking, exercise, stress, and family history.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### `Q2 WHAT IS THE PATIENT AGE DISTRIBUTION?

```python
sns.histplot(data=df, x="age", bins=10, kde=True)
plt.title("Patient Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.show()
```

### The age range with the tallest bars represents the most common patient age group.

### Q3 HOW MANY PATIENTS HAVE HEART DISEASE?

```python
sns.countplot(data=df, x="has_heart_disease")
plt.title("Heart Disease Count (0 = No, 1 = Yes)")
plt.xlabel("Has Heart Disease")
plt.ylabel("Number of Patients")
plt.show()
```

### Q4 how many male and female patients?

```python
sns.countplot(data=df, x="sex")
plt.title("Sex Distribution")
plt.xlabel("Sex")
plt.ylabel("Number of Patients")
plt.show()
```

### Q5 Do patients with heart disease have higher cholesterol?

```python
sns.boxplot(
    data=df,
    x="has_heart_disease",
    y="cholesterol_total"
)
plt.title("Cholesterol vs Heart Disease")
plt.xlabel("Heart Disease")
plt.ylabel("Total Cholesterol")
plt.show()
```

#### The chart compares cholesterol levels between patients with and without heart disease. It shows differences in their median, spread, and possible outliers. This helps identify whether cholesterol may be related to heart disease risk.

### Q6 What is the distribution of systolic blood pressure?

```python
plt.figure(figsize=(8,5))
sns.histplot(df["resting_bp_systolic"], kde=True)
plt.title("Distribution of Systolic Blood Pressure")
plt.xlabel("Systolic Blood Pressure")
plt.ylabel("Number of Patients")
plt.show()
```

#### This graph shows how systolic blood pressure is distributed among the patients. It helps identify the most common blood pressure range and the overall spread of the values. It can also highlight unusually high or low readings.

#### Q7 WHICH SMOKER STATUS HAS THE HIGHEST HEART DISEASE RATE?

```python
smoke_rate = pd.crosstab(
 df["smoker_status"],
 df["has_heart_disease"],
 normalize="index"
) * 100
print(smoke_rate)
smoke_rate.plot(kind="bar")
plt.title("Heart Disease Rate by Smoker Status")
plt.xlabel("Smoker Status")
plt.ylabel("Percentage")
plt.xticks(rotation=0)
plt.show()
```

### Current smokers show a higher "1" percentage,meaning smoking is linked with higher heart disease rate.


### Q8  ARE THERE ANY BMI OUTLIERS?

```python
sns.boxplot(y=df["bmi"])
plt.title("BMI - Outlier Detection")
plt.ylabel("BMI")
plt.show()
```

### Q9 Does family history relate to heart disease?

```python
sns.countplot(data=df,
              x="family_history",
              hue="has_heart_disease")

plt.title("Family History vs Heart Disease")
plt.xlabel("Family History")
plt.ylabel("Number of Patients")
plt.show()
```
